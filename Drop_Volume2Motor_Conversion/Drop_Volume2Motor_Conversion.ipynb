```python
def poly(input,output,x2,y2,DD):
    Values = np.polyfit(input, output, deg= DD)
    p = np.poly1d(Values)
    x_linespace = np.linspace(input.min(),input.max(),int(input.max() - input.min()))
    y_est = p(x_linespace)

    # g, f, e, d, c, b, a, h = np.polyfit(output, output, deg= 7)
    # y_est = (a * output) + (b * output**2) + (c * output**3) + (d * output**4) + (e * output**5) + (f * output**6) + (g * output**7) + h
    # f, e, d, c, b, a, g = np.polyfit(output, output, deg= 6)
    # y_est = (a * output) + (b * output**2) + (c * output**3) + (d * output**4) + (e * output**5) + (f * output**6) + g 

    y_err = x_linespace.std() * np.sqrt(1/len(x_linespace) + (x_linespace - x_linespace.mean())**2 / np.sum((x_linespace - x_linespace.mean())**2))

    Total = 0
    error_Cfit = []
    e_num = 0
    e2 = []
    for i,xx in enumerate(x2):
        Total += 1
        ee = p(xx)- y2[i]
        # ee = (a * yy) + (b * yy**2) + (c * yy**3) + (d * yy**4) + (e * yy**5) + (f * yy**6) + (g * yy**7) + h - x[i]
        # ee = (a * xx) + (b * xx**2) + (c * xx**3) + (d * xx**4) + (e * xx**5) + (f * xx**6) + g - x[i]
        error_Cfit = np.append(error_Cfit,ee)
        if abs(ee)>12:
            print (y[i] , '->' , ee)
            e3 = np.array([x2[i] , y2[i],ee])
            e2.append(e3)
            e_num += 1
    # np.int32(error_Cfit)
    # print('error_Cfit=',error_Cfit)
    print('number of abs sys. errors more than 12=',e_num, "out of:", Total)
    # print(e2)
    e2 = np.array(e2)
    x_e,y_e = e2[:,0] , e2[:,1]
    # fig, ax = plt.subplots()
    # plt.scatter(input, output, color='purple' ,label = "dataset") 
    # plt.figure(figsize=(30,18))
    plt.plot(x_linespace, y_est , label = "predicted using curvefit")#, '-.'
    plt.fill_between(x_linespace, y_est - y_err, y_est + y_err, alpha=0.7,color = 'orange' , label = "error")
    plt.scatter(x2,y2, color='white' , label = "mean")
    plt.scatter(x_e,y_e , label = "error",c=e2[:,2], cmap="bwr")
    plt.colorbar()
    if np.array(x2<4500).all():
        plt.xlabel('uL')  
        plt.ylabel('step')
        plt.xticks()
        plt.savefig('.\\predict steps using inputted uL using curvefit showing disallowed experiments.pdf',format='pdf')
    else:
        plt.xlabel('step')
        plt.ylabel('uL')
        plt.xticks()
        plt.savefig('.\\predict ul using inputted step using curvefit showing disallowed experiments.pdf',format='pdf')
    plt.legend()
    
    plt.show()
    return p , e2 , error_Cfit , Values
```
